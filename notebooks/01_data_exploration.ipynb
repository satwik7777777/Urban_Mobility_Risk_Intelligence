```python
from pathlib import Path
import pandas as pd

# Find all station-level crash files
station_files = [
    file for file in files
    if "Station" in pd.read_csv(file, nrows=0).columns
]

# Collect every unique raw station name
all_station_names = set()

for file in station_files:
    df = pd.read_csv(file)

    stations = (
        df["Station"]
        .dropna()
        .astype(str)
        .str.strip()
    )

    # Remove totals/summary rows
    stations = stations[
        ~stations.str.contains(
            r"total$|^total$",
            case=False,
            regex=True
        )
    ]

    all_station_names.update(stations.unique())

# Create crosswalk
crosswalk = pd.DataFrame({
    "raw_station_name": sorted(all_station_names),
    "canonical_station_name": "",
    "mapping_type": "",
    "confidence": "",
    "notes": ""
})

# Save it
REFERENCE_DATA = Path("../data/reference")
REFERENCE_DATA.mkdir(parents=True, exist_ok=True)

crosswalk.to_csv(
    REFERENCE_DATA / "station_crosswalk.csv",
    index=False
)

print(f"Created crosswalk with {len(crosswalk)} unique raw station names.")
print(f"Saved to: {REFERENCE_DATA / 'station_crosswalk.csv'}")
```

```python
# High-confidence station name standardization
station_mapping = {
    "K.R.Pura": "K R Puram",
    "White Field": "Whitefield",
    "RT Nagar": "R T Nagar",
    "Malleswaram": "Malleshwaram",
    "Chikjala": "Chikkajala",
    "HuliMavu": "Hulimavu",
    "Mico layout": "Micolayout",
    "Chamarajapet": "Chamarajpet",
    "Yelahanka": "Yalahanka"
}

# Apply the mapping
crosswalk["canonical_station_name"] = (
    crosswalk["raw_station_name"]
    .map(station_mapping)
    .fillna(crosswalk["raw_station_name"])
)

crosswalk["mapping_type"] = (
    crosswalk["raw_station_name"]
    .apply(
        lambda x: "standardized"
        if x in station_mapping
        else "unchanged"
    )
)

crosswalk["confidence"] = (
    crosswalk["raw_station_name"]
    .apply(
        lambda x: "HIGH"
        if x in station_mapping
        else "PENDING"
    )
)

crosswalk.to_csv(
    REFERENCE_DATA / "station_crosswalk.csv",
    index=False
)

print("High-confidence mappings applied.")
print(f"Total raw station names: {len(crosswalk)}")
print(f"Standardized: {sum(crosswalk['mapping_type'] == 'standardized')}")
print(f"Pending review: {sum(crosswalk['confidence'] == 'PENDING')}")
```

```python
# Names that need actual human/domain review
review_names = {
    "B.Pura",
    "Y.Pura",
    "K.Swamy Lyt.",
    "Chickpet",
    "Sheshadripuram/Chickpet",
    "K R Puram",
    "K.R.Pura",
    "Mahadevapura",
    "Devanahalli",
    "Bytarayanapura",
    "Yalahanka",
    "Yelahanka",
    "K S Layout",
    "K.Swamy Lyt."
}

# Reset classification
crosswalk["mapping_type"] = "unchanged"
crosswalk["confidence"] = "HIGH"

# Apply our known standardizations
for raw_name, canonical_name in station_mapping.items():
    mask = crosswalk["raw_station_name"] == raw_name
    crosswalk.loc[mask, "canonical_station_name"] = canonical_name
    crosswalk.loc[mask, "mapping_type"] = "standardized"
    crosswalk.loc[mask, "confidence"] = "HIGH"

# Mark ambiguous cases for review
for name in review_names:
    mask = crosswalk["raw_station_name"] == name
    crosswalk.loc[mask, "mapping_type"] = "REVIEW"
    crosswalk.loc[mask, "confidence"] = "PENDING"

# Save
crosswalk.to_csv(
    REFERENCE_DATA / "station_crosswalk.csv",
    index=False
)

print("Classification updated.")
print("\nMapping summary:")
print(crosswalk["mapping_type"].value_counts())
```

```python
from pathlib import Path
import pandas as pd


# --------------------------------------------------
# PATHS
# --------------------------------------------------

PROJECT_ROOT = Path(r"C:\Users\satwi\OneDrive\Desktop\PROJECTS\Urban_Mobility_Risk_Intelligence\DATA FILES")

RAW_DATA = PROJECT_ROOT / "data" / "raw"
PROCESSED_DATA = PROJECT_ROOT / "data" / "processed"

PROCESSED_DATA.mkdir(parents=True, exist_ok=True)


# --------------------------------------------------
# FILES
# --------------------------------------------------

FILES = {
    "2018_2020": RAW_DATA / "2018–2020 station dataset.csv",
    "2021_2022": RAW_DATA / "2020–2022 station dataset.csv",
    "2023": RAW_DATA / "2023 station dataset.csv",
    "2024": RAW_DATA / "2024 station dataset.csv",
    "2025": RAW_DATA / "2025 station dataset.csv",
}


# --------------------------------------------------
# HELPER
# --------------------------------------------------

def remove_total_rows(df):
    """Remove station aggregation/summary rows."""

    station = (
        df["Station"]
        .fillna("")
        .astype(str)
        .str.strip()
    )

    mask = (
        station.str.lower().eq("total")
        | station.str.lower().str.endswith("total")
        | station.str.lower().eq("grand total")
    )

    return df.loc[~mask].copy()


# --------------------------------------------------
# 2018–2020
# --------------------------------------------------

def process_2018_2020():

    df = pd.read_csv(FILES["2018_2020"])

    df = remove_total_rows(df)

    records = []

    for year in [2018, 2019, 2020]:

        records.append(
            pd.DataFrame({
                "zone": df["Zone"],
                "sub_division": pd.NA,
                "station": df["Station"],
                "year": year,
                "fatal_crashes": df[f"{year} - Fatal"].values,
                "nonfatal_crashes": df[f"{year} - Non-Fatal"].values,
                "total_crashes": df[f"{year} - Total Cases"].values,
                "people_killed": df[f"{year} - Killed"].values,
                "people_injured": df[f"{year} - Injured"].values,
            })
        )

    return pd.concat(records, ignore_index=True)


# --------------------------------------------------
# 2021–2022
# --------------------------------------------------

def process_2021_2022():

    df = pd.read_csv(FILES["2021_2022"])

    df = remove_total_rows(df)

    records = []

    for year in [2021, 2022]:

        records.append(
            pd.DataFrame({
                "zone": df["Zone"],
                "sub_division": pd.NA,
                "station": df["Station"],
                "year": year,
                "fatal_crashes": df[f"{year} - Fatal"].values,
                "nonfatal_crashes": df[f"{year} - Non-Fatal"].values,
                "total_crashes": df[f"{year} - Total Cases"].values,
                "people_killed": df[f"{year} - Killed"].values,
                "people_injured": df[f"{year} - Injured"].values,
            })
        )

    return pd.concat(records, ignore_index=True)


# --------------------------------------------------
# 2023
# --------------------------------------------------

def process_2023():

    df = pd.read_csv(FILES["2023"])

    df = remove_total_rows(df)

    return pd.DataFrame({
        "zone": df["Zone"],
        "sub_division": df["Sub-division"],
        "station": df["Station"],
        "year": 2023,
        "fatal_crashes": df["2023 - Fatal Cases"],
        "nonfatal_crashes": df["2023 - Non-Fatal"],
        "total_crashes": df["2023 - Total Cases"],
        "people_killed": df["2023 - Killed People"],
        "people_injured": df["2023 - Injured People"],
    })


# --------------------------------------------------
# 2024
# --------------------------------------------------

def process_2024():

    df = pd.read_csv(FILES["2024"])

    df = remove_total_rows(df)

    return pd.DataFrame({
        "zone": df["Zone"],
        "sub_division": df["Sub-division"],
        "station": df["Station"],
        "year": 2024,
        "fatal_crashes": df["2024-Fatal crashes"],
        "nonfatal_crashes": df["2024-Non Fatal"],
        "total_crashes": df["2024-Total crashes"],
        "people_killed": pd.NA,
        "people_injured": pd.NA,
    })


# --------------------------------------------------
# 2025
# --------------------------------------------------

def process_2025():

    df = pd.read_csv(FILES["2025"])

    df = remove_total_rows(df)

    return pd.DataFrame({
        "zone": df["Zone"],
        "sub_division": df["Sub-division"],
        "station": df["Station"],
        "year": 2025,
        "fatal_crashes": df["2025-Fatal crashes"],
        "nonfatal_crashes": df["2025-Non Fatal"],
        "total_crashes": df["2025-Total crashes"],
        "people_killed": pd.NA,
        "people_injured": pd.NA,
    })


# --------------------------------------------------
# BUILD MASTER DATASET
# --------------------------------------------------

def main():

    print("Processing crash datasets...")

    datasets = [
        process_2018_2020(),
        process_2021_2022(),
        process_2023(),
        process_2024(),
        process_2025(),
    ]

    master = pd.concat(
        datasets,
        ignore_index=True
    )

    # Standardize station names
    station_mapping = {
        "K.R.Pura": "K R Puram",
        "White Field": "Whitefield",
        "RT Nagar": "R T Nagar",
        "Malleswaram": "Malleshwaram",
        "Chikjala": "Chikkajala",
        "HuliMavu": "Hulimavu",
        "Mico layout": "Micolayout",
        "Chamarajapet": "Chamarajpet",
        "Yelahanka": "Yalahanka",
    }

    master["station"] = (
        master["station"]
        .replace(station_mapping)
        .str.strip()
    )

    # Sort
    master = master.sort_values(
        ["year", "zone", "station"]
    ).reset_index(drop=True)

    # Save
    output_file = (
        PROCESSED_DATA /
        "bengaluru_crashes_station_master.csv"
    )

    master.to_csv(
        output_file,
        index=False
    )

    print()
    print("MASTER DATASET CREATED")
    print("-" * 50)
    print(f"Rows: {len(master):,}")
    print(f"Columns: {len(master.columns)}")
    print(f"Years: {master['year'].min()}–{master['year'].max()}")
    print(f"Stations: {master['station'].nunique()}")
    print()
    print(f"Saved to: {output_file}")


if __name__ == "__main__":
    main()
```

```python
from pathlib import Path
import pandas as pd


# ==================================================
# FIND THE FIVE CSV FILES
# ==================================================

print("Files available:")

for file in files:
    print(" -", file.name)


FILES = {}

for file in files:

    name = file.name.lower()

    if "2018" in name:
        FILES["2018_2020"] = file

    elif "2020" in name:
        FILES["2021_2022"] = file

    elif "2023" in name:
        FILES["2023"] = file

    elif "2024" in name:
        FILES["2024"] = file

    elif "2025" in name:
        FILES["2025"] = file


print("\nDatasets identified:")

for key, file in FILES.items():
    print(f"✓ {key}: {file.name}")


# ==================================================
# CHECK ALL DATASETS
# ==================================================

required = [
    "2018_2020",
    "2021_2022",
    "2023",
    "2024",
    "2025"
]

missing = [
    name for name in required
    if name not in FILES
]

if missing:
    raise FileNotFoundError(
        f"Could not identify: {missing}"
    )


# ==================================================
# CSV READER
# ==================================================

def read_csv_clean(file):

    df = pd.read_csv(file)

    # Remove accidental spaces from column names
    df.columns = (
        df.columns
        .astype(str)
        .str.strip()
    )

    return df


# ==================================================
# REMOVE TOTAL / SUMMARY ROWS
# ==================================================

def remove_total_rows(df):

    station = (
        df["Station"]
        .fillna("")
        .astype(str)
        .str.strip()
    )

    remove = (
        station.str.lower().eq("total")
        | station.str.lower().str.endswith("total")
        | station.str.lower().eq("grand total")
    )

    return df.loc[~remove].copy()


# ==================================================
# 2018–2020
# ==================================================

def process_2018_2020():

    df = read_csv_clean(
        FILES["2018_2020"]
    )

    df = remove_total_rows(df)

    output = []

    for year in [2018, 2019, 2020]:

        temp = pd.DataFrame({

            "zone":
                df["Zone"].values,

            "sub_division":
                pd.NA,

            "station":
                df["Station"].values,

            "year":
                year,

            "fatal_crashes":
                df[f"{year} - Fatal"].values,

            "nonfatal_crashes":
                df[f"{year} - Non-Fatal"].values,

            "total_crashes":
                df[f"{year} - Total Cases"].values,

            "people_killed":
                df[f"{year} - Killed"].values,

            "people_injured":
                df[f"{year} - Injured"].values
        })

        output.append(temp)

    return pd.concat(
        output,
        ignore_index=True
    )


# ==================================================
# 2021–2022
# ==================================================

def process_2021_2022():

    df = read_csv_clean(
        FILES["2021_2022"]
    )

    df = remove_total_rows(df)

    output = []

    for year in [2021, 2022]:

        temp = pd.DataFrame({

            "zone":
                df["Zone"].values,

            "sub_division":
                pd.NA,

            "station":
                df["Station"].values,

            "year":
                year,

            "fatal_crashes":
                df[f"{year} - Fatal"].values,

            "nonfatal_crashes":
                df[f"{year} - Non-Fatal"].values,

            "total_crashes":
                df[f"{year} - Total Cases"].values,

            "people_killed":
                df[f"{year} - Killed"].values,

            "people_injured":
                df[f"{year} - Injured"].values
        })

        output.append(temp)

    return pd.concat(
        output,
        ignore_index=True
    )


# ==================================================
# 2023
# ==================================================

def process_2023():

    df = read_csv_clean(
        FILES["2023"]
    )

    df = remove_total_rows(df)

    return pd.DataFrame({

        "zone":
            df["Zone"].values,

        "sub_division":
            df["Sub-division"].values,

        "station":
            df["Station"].values,

        "year":
            2023,

        "fatal_crashes":
            df["2023 - Fatal Cases"].values,

        "nonfatal_crashes":
            df["2023 - Non-Fatal"].values,

        "total_crashes":
            df["2023 - Total Cases"].values,

        "people_killed":
            df["2023 - Killed People"].values,

        "people_injured":
            df["2023 - Injured People"].values
    })


# ==================================================
# 2024
# ==================================================

def process_2024():

    df = read_csv_clean(
        FILES["2024"]
    )

    df = remove_total_rows(df)

    return pd.DataFrame({

        "zone":
            df["Zone"].values,

        "sub_division":
            df["Sub-division"].values,

        "station":
            df["Station"].values,

        "year":
            2024,

        "fatal_crashes":
            df["2024-Fatal crashes"].values,

        "nonfatal_crashes":
            df["2024-Non Fatal"].values,

        "total_crashes":
            df["2024-Total crashes"].values,

        # Not provided in 2024 dataset
        "people_killed":
            pd.NA,

        "people_injured":
            pd.NA
    })


# ==================================================
# 2025
# ==================================================

def process_2025():

    df = read_csv_clean(
        FILES["2025"]
    )

    df = remove_total_rows(df)

    return pd.DataFrame({

        "zone":
            df["Zone"].values,

        "sub_division":
            df["Sub-division"].values,

        "station":
            df["Station"].values,

        "year":
            2025,

        "fatal_crashes":
            df["2025-Fatal crashes"].values,

        "nonfatal_crashes":
            df["2025-Non Fatal"].values,

        "total_crashes":
            df["2025-Total crashes"].values,

        # Not provided in 2025 dataset
        "people_killed":
            pd.NA,

        "people_injured":
            pd.NA
    })


# ==================================================
# PROCESS ALL DATASETS
# ==================================================

print("\nProcessing datasets...")

data_2018_2020 = process_2018_2020()

print(
    f"2018–2020 processed: "
    f"{len(data_2018_2020)} rows"
)

data_2021_2022 = process_2021_2022()

print(
    f"2021–2022 processed: "
    f"{len(data_2021_2022)} rows"
)

data_2023 = process_2023()

print(
    f"2023 processed: "
    f"{len(data_2023)} rows"
)

data_2024 = process_2024()

print(
    f"2024 processed: "
    f"{len(data_2024)} rows"
)

data_2025 = process_2025()

print(
    f"2025 processed: "
    f"{len(data_2025)} rows"
)


# ==================================================
# COMBINE EVERYTHING
# ==================================================

master = pd.concat(
    [
        data_2018_2020,
        data_2021_2022,
        data_2023,
        data_2024,
        data_2025
    ],
    ignore_index=True
)


# ==================================================
# STANDARDIZE OBVIOUS STATION NAME VARIATIONS
# ==================================================

station_mapping = {

    "K.R.Pura":
        "K R Puram",

    "White Field":
        "Whitefield",

    "RT Nagar":
        "R T Nagar",

    "Malleswaram":
        "Malleshwaram",

    "Chikjala":
        "Chikkajala",

    "HuliMavu":
        "Hulimavu",

    "Mico layout":
        "Micolayout",

    "Chamarajapet":
        "Chamarajpet",

    "Yelahanka":
        "Yalahanka"
}


master["station"] = (
    master["station"]
    .replace(station_mapping)
    .str.strip()
)


# ==================================================
# DATA TYPES
# ==================================================

master["year"] = (
    master["year"]
    .astype(int)
)


numeric_columns = [
    "fatal_crashes",
    "nonfatal_crashes",
    "total_crashes",
    "people_killed",
    "people_injured"
]


for column in numeric_columns:

    master[column] = pd.to_numeric(
        master[column],
        errors="coerce"
    )


# ==================================================
# SORT
# ==================================================

master = master.sort_values(
    [
        "year",
        "zone",
        "station"
    ]
).reset_index(drop=True)


# ==================================================
# OUTPUT PATH
# ==================================================

OUTPUT_FOLDER = (
    Path(files[0]).parent.parent /
    "processed"
)

OUTPUT_FOLDER.mkdir(
    parents=True,
    exist_ok=True
)


output_file = (
    OUTPUT_FOLDER /
    "bengaluru_crashes_station_master.csv"
)


# ==================================================
# SAVE
# ==================================================

master.to_csv(
    output_file,
    index=False
)


# ==================================================
# FINAL SUMMARY
# ==================================================

print("\n")
print("=" * 60)
print("MASTER DATASET CREATED SUCCESSFULLY")
print("=" * 60)

print(
    f"Rows: {len(master):,}"
)

print(
    f"Columns: {len(master.columns)}"
)

print(
    f"Years: "
    f"{master['year'].min()}–"
    f"{master['year'].max()}"
)

print(
    f"Unique station names: "
    f"{master['station'].nunique()}"
)


print("\nRows by year:")

print(
    master.groupby("year").size()
)


print("\nMissing values:")

print(
    master.isna().sum()
)


print("\nSaved to:")

print(output_file)
```

```python
# ============================================================
# STATION RISK FEATURES
# ============================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("CREATING STATION RISK FEATURES")
print("=" * 70)

# ------------------------------------------------------------
# 1. BASIC STATION METRICS
# ------------------------------------------------------------

risk_features = (
    master
    .groupby("station")
    .agg(
        years_available=("year", "nunique"),
        first_year=("year", "min"),
        last_year=("year", "max"),

        total_crashes=("total_crashes", "sum"),
        total_fatal_crashes=("fatal_crashes", "sum"),

        total_people_killed=("people_killed", "sum"),
        total_people_injured=("people_injured", "sum")
    )
    .reset_index()
)

# Average annual crashes
risk_features["avg_annual_crashes"] = (
    risk_features["total_crashes"]
    / risk_features["years_available"]
)

# Average annual fatal crashes
risk_features["avg_annual_fatal_crashes"] = (
    risk_features["total_fatal_crashes"]
    / risk_features["years_available"]
)

# Fatal crash percentage
risk_features["fatal_crash_pct"] = (
    risk_features["total_fatal_crashes"]
    / risk_features["total_crashes"]
) * 100


# ------------------------------------------------------------
# 2. RECENT PERFORMANCE
# ------------------------------------------------------------

recent_data = master[
    master["year"].between(2023, 2025)
].copy()

recent_features = (
    recent_data
    .groupby("station")
    .agg(
        recent_years_available=("year", "nunique"),
        recent_total_crashes=("total_crashes", "sum"),
        recent_fatal_crashes=("fatal_crashes", "sum")
    )
    .reset_index()
)

# Recent average annual crashes
recent_features["recent_avg_annual_crashes"] = (
    recent_features["recent_total_crashes"]
    / recent_features["recent_years_available"]
)

# Recent average annual fatal crashes
recent_features["recent_avg_annual_fatal_crashes"] = (
    recent_features["recent_fatal_crashes"]
    / recent_features["recent_years_available"]
)

# Recent fatal crash percentage
recent_features["recent_fatal_crash_pct"] = (
    recent_features["recent_fatal_crashes"]
    / recent_features["recent_total_crashes"]
) * 100


# ------------------------------------------------------------
# 3. MERGE RECENT FEATURES
# ------------------------------------------------------------

risk_features = risk_features.merge(
    recent_features,
    on="station",
    how="left"
)


# ------------------------------------------------------------
# 4. CRASH TREND
# ------------------------------------------------------------

trend_results = []

for station, group in master.groupby("station"):

    group = group.sort_values("year")

    # Need at least 2 years to calculate a trend
    if len(group) >= 2:

        slope = np.polyfit(
            group["year"],
            group["total_crashes"],
            1
        )[0]

    else:
        slope = np.nan

    trend_results.append({
        "station": station,
        "crash_trend_per_year": slope
    })

trend_features = pd.DataFrame(trend_results)

risk_features = risk_features.merge(
    trend_features,
    on="station",
    how="left"
)


# ------------------------------------------------------------
# 5. RECENT VS HISTORICAL CHANGE
# ------------------------------------------------------------

risk_features["recent_vs_historical_pct"] = (
    (
        risk_features["recent_avg_annual_crashes"]
        - risk_features["avg_annual_crashes"]
    )
    / risk_features["avg_annual_crashes"]
) * 100


# ------------------------------------------------------------
# 6. ROUND VALUES
# ------------------------------------------------------------

numeric_columns = [
    "avg_annual_crashes",
    "avg_annual_fatal_crashes",
    "fatal_crash_pct",
    "recent_avg_annual_crashes",
    "recent_avg_annual_fatal_crashes",
    "recent_fatal_crash_pct",
    "crash_trend_per_year",
    "recent_vs_historical_pct"
]

for column in numeric_columns:
    risk_features[column] = risk_features[column].round(2)


# ------------------------------------------------------------
# 7. SORT BY AVERAGE ANNUAL CRASHES
# ------------------------------------------------------------

risk_features = risk_features.sort_values(
    "avg_annual_crashes",
    ascending=False
).reset_index(drop=True)


# ------------------------------------------------------------
# 8. DISPLAY RESULTS
# ------------------------------------------------------------

print("\nRisk features created successfully.")

print("\nNumber of stations:", len(risk_features))

print("\nTop stations by average annual crashes:")

print(
    risk_features[
        [
            "station",
            "years_available",
            "avg_annual_crashes",
            "avg_annual_fatal_crashes",
            "fatal_crash_pct",
            "recent_avg_annual_crashes",
            "recent_avg_annual_fatal_crashes",
            "crash_trend_per_year"
        ]
    ]
    .head(20)
    .to_string(index=False)
)


# ------------------------------------------------------------
# 9. SAVE DATASET
# ------------------------------------------------------------

risk_file = OUTPUT_FOLDER / "station_risk_features.csv"

risk_features.to_csv(
    risk_file,
    index=False
)

print("\n")
print("=" * 70)
print("RISK FEATURES DATASET CREATED")
print("=" * 70)

print("\nSaved to:")
print(risk_file)
```

```python
# ============================================================
# STATION RISK SCORE
# ============================================================

print("=" * 70)
print("CALCULATING STATION RISK SCORE")
print("=" * 70)

# ------------------------------------------------------------
# 1. CREATE A COPY
# ------------------------------------------------------------

risk_scored = risk_features.copy()


# ------------------------------------------------------------
# 2. NORMALIZE THE MAIN RISK INDICATORS
# ------------------------------------------------------------

def min_max_score(series):
    minimum = series.min()
    maximum = series.max()

    if maximum == minimum:
        return pd.Series(50, index=series.index)

    return (
        (series - minimum)
        / (maximum - minimum)
    ) * 100


# Crash frequency score
risk_scored["crash_frequency_score"] = min_max_score(
    risk_scored["avg_annual_crashes"]
)

# Fatal crash score
risk_scored["fatality_score"] = min_max_score(
    risk_scored["avg_annual_fatal_crashes"]
)

# Fatal crash percentage score
risk_scored["severity_score"] = min_max_score(
    risk_scored["fatal_crash_pct"]
)

# Recent crash score
risk_scored["recent_crash_score"] = min_max_score(
    risk_scored["recent_avg_annual_crashes"]
)

# Recent fatal crash score
risk_scored["recent_fatality_score"] = min_max_score(
    risk_scored["recent_avg_annual_fatal_crashes"]
)


# ------------------------------------------------------------
# 3. CALCULATE OVERALL RISK SCORE
# ------------------------------------------------------------

risk_scored["risk_score"] = (
    risk_scored["crash_frequency_score"] * 0.30
    + risk_scored["fatality_score"] * 0.25
    + risk_scored["severity_score"] * 0.15
    + risk_scored["recent_crash_score"] * 0.20
    + risk_scored["recent_fatality_score"] * 0.10
)


# ------------------------------------------------------------
# 4. ROUND SCORE
# ------------------------------------------------------------

risk_scored["risk_score"] = (
    risk_scored["risk_score"]
    .round(2)
)


# ------------------------------------------------------------
# 5. CREATE RISK CATEGORY
# ------------------------------------------------------------

def classify_risk(score):

    if score >= 70:
        return "High"

    elif score >= 40:
        return "Medium"

    else:
        return "Low"


risk_scored["risk_category"] = (
    risk_scored["risk_score"]
    .apply(classify_risk)
)


# ------------------------------------------------------------
# 6. DATA QUALITY FLAG
# ------------------------------------------------------------

def coverage_flag(years):

    if years >= 8:
        return "Strong"

    elif years >= 5:
        return "Moderate"

    else:
        return "Limited"


risk_scored["data_coverage"] = (
    risk_scored["years_available"]
    .apply(coverage_flag)
)


# ------------------------------------------------------------
# 7. FINAL RANKING
# ------------------------------------------------------------

risk_scored = risk_scored.sort_values(
    "risk_score",
    ascending=False
).reset_index(drop=True)

risk_scored["risk_rank"] = (
    risk_scored.index + 1
)


# ------------------------------------------------------------
# 8. DISPLAY TOP 20
# ------------------------------------------------------------

print("\nTop 20 highest-risk stations:\n")

print(
    risk_scored[
        [
            "risk_rank",
            "station",
            "risk_score",
            "risk_category",
            "data_coverage",
            "years_available",
            "avg_annual_crashes",
            "avg_annual_fatal_crashes",
            "fatal_crash_pct",
            "recent_avg_annual_crashes"
        ]
    ]
    .head(20)
    .to_string(index=False)
)


# ------------------------------------------------------------
# 9. RISK CATEGORY SUMMARY
# ------------------------------------------------------------

print("\n")
print("=" * 70)
print("RISK CATEGORY SUMMARY")
print("=" * 70)

print(
    risk_scored["risk_category"]
    .value_counts()
)


# ------------------------------------------------------------
# 10. SAVE FINAL RISK DATASET
# ------------------------------------------------------------

risk_score_file = (
    OUTPUT_FOLDER /
    "station_risk_score.csv"
)

risk_scored.to_csv(
    risk_score_file,
    index=False
)


print("\n")
print("=" * 70)
print("FINAL RISK SCORE DATASET CREATED")
print("=" * 70)

print("\nSaved to:")
print(risk_score_file)
```

```python
# ============================================================
# CORRECTED STATION RISK SCORE
# ============================================================

print("=" * 70)
print("RECALCULATING RISK SCORE")
print("=" * 70)

risk_scored = risk_features.copy()


# ------------------------------------------------------------
# 1. NORMALIZATION FUNCTION
# ------------------------------------------------------------

def min_max_score(series):

    minimum = series.min()
    maximum = series.max()

    if maximum == minimum:
        return pd.Series(50.0, index=series.index)

    return (
        (series - minimum)
        / (maximum - minimum)
    ) * 100


# ------------------------------------------------------------
# 2. HISTORICAL RISK COMPONENTS
# ------------------------------------------------------------

risk_scored["crash_frequency_score"] = min_max_score(
    risk_scored["avg_annual_crashes"]
)

risk_scored["fatality_score"] = min_max_score(
    risk_scored["avg_annual_fatal_crashes"]
)

risk_scored["severity_score"] = min_max_score(
    risk_scored["fatal_crash_pct"]
)


# ------------------------------------------------------------
# 3. RECENT RISK COMPONENTS
# ------------------------------------------------------------

risk_scored["recent_crash_score"] = min_max_score(
    risk_scored["recent_avg_annual_crashes"]
)

risk_scored["recent_fatality_score"] = min_max_score(
    risk_scored["recent_avg_annual_fatal_crashes"]
)


# ------------------------------------------------------------
# 4. CALCULATE SCORE USING AVAILABLE DATA
# ------------------------------------------------------------

risk_scored["risk_score"] = np.nan

for index, row in risk_scored.iterrows():

    historical_score = (
        row["crash_frequency_score"] * 0.40
        + row["fatality_score"] * 0.30
        + row["severity_score"] * 0.30
    )

    # If recent data exists
    if pd.notna(row["recent_crash_score"]) and pd.notna(
        row["recent_fatality_score"]
    ):

        risk_score = (
            row["crash_frequency_score"] * 0.25
            + row["fatality_score"] * 0.20
            + row["severity_score"] * 0.15
            + row["recent_crash_score"] * 0.25
            + row["recent_fatality_score"] * 0.15
        )

    # If recent data does not exist
    else:

        risk_score = historical_score

    risk_scored.loc[index, "risk_score"] = risk_score


# ------------------------------------------------------------
# 5. ROUND SCORE
# ------------------------------------------------------------

risk_scored["risk_score"] = (
    risk_scored["risk_score"]
    .round(2)
)


# ------------------------------------------------------------
# 6. RISK CATEGORY
# ------------------------------------------------------------

def classify_risk(score):

    if score >= 70:
        return "High"

    elif score >= 40:
        return "Medium"

    else:
        return "Low"


risk_scored["risk_category"] = (
    risk_scored["risk_score"]
    .apply(classify_risk)
)


# ------------------------------------------------------------
# 7. DATA COVERAGE
# ------------------------------------------------------------

def coverage_flag(years):

    if years >= 8:
        return "Strong"

    elif years >= 5:
        return "Moderate"

    else:
        return "Limited"


risk_scored["data_coverage"] = (
    risk_scored["years_available"]
    .apply(coverage_flag)
)


# ------------------------------------------------------------
# 8. RECENT DATA AVAILABILITY
# ------------------------------------------------------------

risk_scored["recent_data_status"] = np.where(
    risk_scored["recent_years_available"] >= 3,
    "Complete",
    np.where(
        risk_scored["recent_years_available"] > 0,
        "Partial",
        "Unavailable"
    )
)


# ------------------------------------------------------------
# 9. RANK STATIONS
# ------------------------------------------------------------

risk_scored = risk_scored.sort_values(
    "risk_score",
    ascending=False
).reset_index(drop=True)

risk_scored["risk_rank"] = (
    risk_scored.index + 1
)


# ------------------------------------------------------------
# 10. DISPLAY TOP 20
# ------------------------------------------------------------

print("\nTop 20 stations by corrected risk score:\n")

print(
    risk_scored[
        [
            "risk_rank",
            "station",
            "risk_score",
            "risk_category",
            "data_coverage",
            "recent_data_status",
            "years_available",
            "avg_annual_crashes",
            "avg_annual_fatal_crashes",
            "fatal_crash_pct"
        ]
    ]
    .head(20)
    .to_string(index=False)
)


# ------------------------------------------------------------
# 11. CATEGORY SUMMARY
# ------------------------------------------------------------

print("\n")
print("=" * 70)
print("RISK CATEGORY SUMMARY")
print("=" * 70)

print(
    risk_scored["risk_category"]
    .value_counts()
    .reindex(["High", "Medium", "Low"])
    .fillna(0)
    .astype(int)
)


# ------------------------------------------------------------
# 12. SAVE CORRECTED DATASET
# ------------------------------------------------------------

risk_score_file = (
    OUTPUT_FOLDER /
    "station_risk_score.csv"
)

risk_scored.to_csv(
    risk_score_file,
    index=False
)

print("\n")
print("=" * 70)
print("CORRECTED RISK SCORE DATASET SAVED")
print("=" * 70)

print("\nSaved to:")
print(risk_score_file)
```

```python
# ============================================================
# STATION-LEVEL CRASH TREND ANALYSIS
# ============================================================

print("=" * 70)
print("STATION-LEVEL CRASH TREND ANALYSIS")
print("=" * 70)

# ------------------------------------------------------------
# 1. CALCULATE TREND FOR EACH STATION
# ------------------------------------------------------------

station_trends = []

for station, group in master.groupby("station"):

    group = group.sort_values("year")

    # Need at least 2 years
    if len(group) >= 2:

        slope = np.polyfit(
            group["year"],
            group["total_crashes"],
            1
        )[0]

        first_crashes = group.iloc[0]["total_crashes"]
        last_crashes = group.iloc[-1]["total_crashes"]

        # Percentage change from first available year
        if first_crashes > 0:
            change_pct = (
                (last_crashes - first_crashes)
                / first_crashes
            ) * 100
        else:
            change_pct = np.nan

    else:
        slope = np.nan
        change_pct = np.nan

    station_trends.append({
        "station": station,
        "crash_trend_per_year": slope,
        "first_vs_last_change_pct": change_pct
    })


station_trends = pd.DataFrame(station_trends)


# ------------------------------------------------------------
# 2. MERGE WITH RISK DATA
# ------------------------------------------------------------

station_trends = station_trends.merge(
    risk_scored[
        [
            "station",
            "risk_score",
            "risk_category",
            "years_available",
            "data_coverage"
        ]
    ],
    on="station",
    how="left"
)


# ------------------------------------------------------------
# 3. ROUND VALUES
# ------------------------------------------------------------

station_trends["crash_trend_per_year"] = (
    station_trends["crash_trend_per_year"]
    .round(2)
)

station_trends["first_vs_last_change_pct"] = (
    station_trends["first_vs_last_change_pct"]
    .round(2)
)


# ------------------------------------------------------------
# 4. STATIONS WITH INCREASING CRASH TREND
# ------------------------------------------------------------

increasing = (
    station_trends
    .sort_values(
        "crash_trend_per_year",
        ascending=False
    )
)

print("\nTop 15 stations with increasing crash trends:\n")

print(
    increasing[
        [
            "station",
            "crash_trend_per_year",
            "first_vs_last_change_pct",
            "risk_score",
            "risk_category",
            "years_available",
            "data_coverage"
        ]
    ]
    .head(15)
    .to_string(index=False)
)


# ------------------------------------------------------------
# 5. STATIONS WITH DECREASING CRASH TREND
# ------------------------------------------------------------

decreasing = (
    station_trends
    .sort_values(
        "crash_trend_per_year",
        ascending=True
    )
)

print("\n")
print("=" * 70)
print("TOP 15 STATIONS WITH DECREASING CRASH TRENDS")
print("=" * 70)

print(
    decreasing[
        [
            "station",
            "crash_trend_per_year",
            "first_vs_last_change_pct",
            "risk_score",
            "risk_category",
            "years_available",
            "data_coverage"
        ]
    ]
    .head(15)
    .to_string(index=False)
)


# ------------------------------------------------------------
# 6. SAVE STATION TREND DATA
# ------------------------------------------------------------

station_trend_file = (
    OUTPUT_FOLDER /
    "station_crash_trends.csv"
)

station_trends.to_csv(
    station_trend_file,
    index=False
)


print("\n")
print("=" * 70)
print("STATION TREND DATASET CREATED")
print("=" * 70)

print("\nSaved to:")
print(station_trend_file)
```

```python
# ============================================================
# FINAL STATION RISK PROFILE
# ============================================================

print("=" * 70)
print("CREATING FINAL STATION RISK PROFILE")
print("=" * 70)

# ------------------------------------------------------------
# 1. SELECT IMPORTANT COLUMNS
# ------------------------------------------------------------

station_risk_profile = risk_scored[
    [
        "risk_rank",
        "station",
        "risk_score",
        "risk_category",
        "data_coverage",
        "recent_data_status",
        "years_available",
        "first_year",
        "last_year",
        "avg_annual_crashes",
        "avg_annual_fatal_crashes",
        "fatal_crash_pct",
        "recent_avg_annual_crashes",
        "recent_avg_annual_fatal_crashes",
        "crash_trend_per_year"
    ]
].copy()


# ------------------------------------------------------------
# 2. ADD TREND CATEGORY
# ------------------------------------------------------------

def trend_category(value):

    if pd.isna(value):
        return "Unknown"

    elif value >= 5:
        return "Strongly Increasing"

    elif value > 0:
        return "Increasing"

    elif value <= -5:
        return "Strongly Decreasing"

    else:
        return "Stable / Slightly Decreasing"


station_risk_profile["trend_category"] = (
    station_risk_profile["crash_trend_per_year"]
    .apply(trend_category)
)


# ------------------------------------------------------------
# 3. CREATE PRIORITY FLAG
# ------------------------------------------------------------

def priority_flag(row):

    if (
        row["risk_category"] == "High"
        and row["data_coverage"] == "Strong"
    ):
        return "High Priority"

    elif (
        row["risk_category"] == "High"
        and row["data_coverage"] != "Strong"
    ):
        return "High Risk - Limited Data"

    elif (
        row["risk_category"] == "Medium"
        and row["trend_category"] == "Strongly Increasing"
    ):
        return "Monitor Closely"

    else:
        return "Normal Monitoring"


station_risk_profile["priority_flag"] = (
    station_risk_profile.apply(
        priority_flag,
        axis=1
    )
)


# ------------------------------------------------------------
# 4. SORT BY RISK RANK
# ------------------------------------------------------------

station_risk_profile = (
    station_risk_profile
    .sort_values("risk_rank")
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 5. DISPLAY TOP 20
# ------------------------------------------------------------

print("\nTop 20 station risk profiles:\n")

print(
    station_risk_profile
    .head(20)
    .to_string(index=False)
)


# ------------------------------------------------------------
# 6. PRIORITY SUMMARY
# ------------------------------------------------------------

print("\n")
print("=" * 70)
print("PRIORITY SUMMARY")
print("=" * 70)

print(
    station_risk_profile["priority_flag"]
    .value_counts()
)


# ------------------------------------------------------------
# 7. TREND SUMMARY
# ------------------------------------------------------------

print("\n")
print("=" * 70)
print("TREND SUMMARY")
print("=" * 70)

print(
    station_risk_profile["trend_category"]
    .value_counts()
)


# ------------------------------------------------------------
# 8. SAVE FINAL PROFILE
# ------------------------------------------------------------

profile_file = (
    OUTPUT_FOLDER /
    "station_risk_profile.csv"
)

station_risk_profile.to_csv(
    profile_file,
    index=False
)


print("\n")
print("=" * 70)
print("FINAL STATION RISK PROFILE CREATED")
print("=" * 70)

print("\nSaved to:")
print(profile_file)
```

```python
# ============================================================
# GEOCODE BENGALURU STATIONS
# ============================================================

from geopy.geocoders import Nominatim
from time import sleep
import pandas as pd

print("=" * 70)
print("GEOCODING BENGALURU STATIONS")
print("=" * 70)

# ------------------------------------------------------------
# 1. CREATE GEOCODER
# ------------------------------------------------------------

geolocator = Nominatim(
    user_agent="urban_mobility_risk_intelligence"
)


# ------------------------------------------------------------
# 2. COPY MAP DATA
# ------------------------------------------------------------

geocoded_data = map_data.copy()


# ------------------------------------------------------------
# 3. GEOCODE EACH STATION
# ------------------------------------------------------------

for index, row in geocoded_data.iterrows():

    station = row["station"]

    # Search specifically within Bengaluru
    query = f"{station} Police Station, Bengaluru, Karnataka, India"

    try:

        location = geolocator.geocode(
            query,
            exactly_one=True,
            timeout=10
        )

        if location:

            geocoded_data.loc[index, "latitude"] = (
                location.latitude
            )

            geocoded_data.loc[index, "longitude"] = (
                location.longitude
            )

            geocoded_data.loc[index, "location_status"] = (
                "Geocoded"
            )

            print(
                f"✓ {station}: "
                f"{location.latitude:.6f}, "
                f"{location.longitude:.6f}"
            )

        else:

            geocoded_data.loc[index, "location_status"] = (
                "Not Found"
            )

            print(f"⚠ {station}: Not Found")

    except Exception as error:

        geocoded_data.loc[index, "location_status"] = (
            "Error"
        )

        print(
            f"✗ {station}: {error}"
        )

    # Respect Nominatim usage policy
    sleep(1)


# ------------------------------------------------------------
# 4. CONVERT COORDINATES TO NUMERIC
# ------------------------------------------------------------

geocoded_data["latitude"] = pd.to_numeric(
    geocoded_data["latitude"],
    errors="coerce"
)

geocoded_data["longitude"] = pd.to_numeric(
    geocoded_data["longitude"],
    errors="coerce"
)


# ------------------------------------------------------------
# 5. SUMMARY
# ------------------------------------------------------------

print("\n")
print("=" * 70)
print("GEOCODING SUMMARY")
print("=" * 70)

print(
    geocoded_data["location_status"]
    .value_counts()
)


# ------------------------------------------------------------
# 6. SHOW UNRESOLVED STATIONS
# ------------------------------------------------------------

unresolved = geocoded_data[
    geocoded_data["location_status"] != "Geocoded"
]

print("\n")
print("=" * 70)
print("STATIONS REQUIRING MANUAL REVIEW")
print("=" * 70)

if len(unresolved) == 0:

    print("\nAll stations were geocoded successfully.")

else:

    print(
        unresolved[
            [
                "station",
                "risk_score",
                "risk_category",
                "location_status"
            ]
        ].to_string(index=False)
    )


# ------------------------------------------------------------
# 7. SAVE GEOCODED DATA
# ------------------------------------------------------------

geocoded_file = (
    OUTPUT_FOLDER /
    "station_risk_geocoded.csv"
)

geocoded_data.to_csv(
    geocoded_file,
    index=False
)


print("\n")
print("=" * 70)
print("GEOCODED DATASET CREATED")
print("=" * 70)

print("\nSaved to:")
print(geocoded_file)
```
